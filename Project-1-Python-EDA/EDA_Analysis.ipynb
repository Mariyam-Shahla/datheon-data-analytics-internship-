```python
df['Order Date']=pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date']=pd.to_datetime(df['Ship Date'], errors='coerce')
```

```python
# Sales by Category

category_sales=df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(x=category_sales.index, y=category_sales.values)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
# Profit by Category

category_profit=df.groupby("Category")["Profit"].sum().sort_values(ascending= False)

plt.figure(figsize=(10, 6))
sns.barplot(x=category_profit.index, y=category_profit.values)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
# Monthly Sales Trend

df["Order Date"]= pd.to_datetime(df["Order Date"])

monthly_sales=df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()

plt.figure(figsize=(14, 6))
plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker='o')

plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Sales by Region

region_sales=df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=region_sales.index, y=region_sales.values)

plt.title("Sales by Region", fontsize=16)
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

```python
# Profit Distribution

plt.figure(figsize=(10,6))

sns.histplot(df["Profit"], bins=40, kde=True)

plt.title("Profit Distribution", fontsize=16)
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.show()
```

```python
# Sales vs Profit Relationship

plt.figure(figsize=(10,6))

sns.scatterplot(data=df, x="Sales", y="Profit", hue="Category", alpha=0.7)

plt.title("Sales vs Profit",fontsize=16)
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
           
```

```python
# Sales by Ship Mode

ship_sales=df.groupby("Ship Mode")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=ship_sales.index, y=ship_sales.values)

plt.title("Sales bu Shipping Mode", fontsize=16)
plt.xlabel("Shipping Mode")
plt.ylabel("Total Sales")

plt.show()
```

```python
# Correlation Heatmap

plt.figure(figsize=(10,6))

numeric_df=df.select_dtypes(include= "number")
correlation=numeric_df.corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt=".2f")

plt.title("Correlation Heatmap", fontsize=16)
plt.show()
```

```python
# Sales by Category

subcategory_sales=(df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).head(10))

plt.figure(figsize=(12,6))
sns.barplot(x=subcategory_sales.values, y=subcategory_sales.index)

plt.title("Top 10 Sub-Categories by Sales", fontsize=16)
plt.xlabel("Total Sales")
plt.ylabel("Sub-Category")

plt.show()
```
